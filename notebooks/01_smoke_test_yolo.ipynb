```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

results = model.predict(
    source="https://ultralytics.com/images/bus.jpg",
    conf=0.25,
    save=True
)

print("YOLO smoke test completed")
print(f"Number of result objects: {len(results)}")
```
