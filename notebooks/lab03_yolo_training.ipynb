```python
import os
from pathlib import Path
from ultralytics import YOLO

# notebook лежить у notebooks/, переходимо в корінь проєкту
os.chdir("..")
print(Path.cwd())
```

```python
DATA_PATH = Path("datasets/construction-ppe-small/data.yaml")
PROJECT_DIR = str(Path("runs/lab03_notebook").absolute())
EXPERIMENT_NAME = "yolo11n_construction_ppe_small"

print("Dataset config exists:", DATA_PATH.exists())
```

```python
model = YOLO("yolo11n.pt")
```

```python
results = model.train(
    data=str(DATA_PATH),
    epochs=10,
    imgsz=640,
    batch=4,
    device="cpu",
    project=PROJECT_DIR,
    name=EXPERIMENT_NAME,
    exist_ok=True
)
```

```python
run_dir = Path(PROJECT_DIR) / EXPERIMENT_NAME
print("Run directory:", run_dir)
print("Files:")
for item in sorted(run_dir.iterdir()):
    print(item)
```

```python
weights_dir = run_dir / "weights"
print("Weights:")
for item in weights_dir.iterdir():
    print(item)
```
